```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data_folder = 'Dataset/'

# Normal CSV
df1 = pd.read_csv(data_folder + "wine_1.csv")
# wine_2 
df2 = pd.read_csv(data_folder + "wine_2.csv", skiprows=1)
# wine_3 
df3 = pd.read_csv(data_folder + "wine_3.csv", header=0, nrows=5)
# wine_4
df4 = pd.read_csv(data_folder + "wine_4.txt", sep='\t')
# wine_5 
df5 = pd.read_csv(data_folder + "wine_5.csv")
df5.columns = df5.columns.str.strip()      
df5 = df5[df1.columns]                     
# wine_6 messy file
df1 = pd.read_csv(data_folder + "wine_1.csv")
# Process messy wine_6.txt >:(
with open(data_folder + "wine_6.txt", 'r', encoding='utf-8') as f:
    lines = f.readlines()
    data_lines = lines[2:7]

data_rows = [line.strip().split('\t')[1:8] for line in data_lines]

df6 = pd.DataFrame(data_rows, columns=df1.columns)

numeric_cols = ['Bottle', 'Alcohol', 'pH']
df6[numeric_cols] = df6[numeric_cols].apply(pd.to_numeric)

```

```python
for i, df in enumerate([df2, df3, df4, df5, df6]):
    assert df1.equals(df), f"df1 not equal to df{i + 2}"
print("All tests passed.")
```
